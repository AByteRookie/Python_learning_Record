```python
# 绘制折线图
import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体
1# windows系统
# 创建一个图标，设置大小
plt.figure(figsize=(10, 5))

##添加数据
month = ['1月','2月','3月','4月']
sales = [150,230,180,240]
#绘制折线图
plt.plot(month,sales,
         label='产品A',
         color='orange',
         linewidth=2,
         linestyle='--',
         marker='o')
#添加标题
plt.title('2025销售趋势', color='black',fontsize=20)
# 添加坐标轴标签
plt.xlabel('季度',fontsize=40)
plt.ylabel('销售额（万元）',fontsize=15)
# 添加图例
plt.legend(loc='upper left')
##添加网格线
plt.grid(True,alpha=0.3,color='green',linestyle='--') # 透明度 颜色 虚实线
# plt.grid(axis='y') 只有横格
# plt.grid(axis='x) 只有竖格
# 设置刻度字体大小
plt.xticks(rotation=10,fontsize=20) #旋转角度
plt.yticks(rotation=0,fontsize=15)
# 设置y轴的范围
plt.ylim(100,300)
# 数据点显示数值
for x,y in zip(month,sales):
    plt.text(x,y,str(y),ha='center',va='bottom',fontsize=13) # 水平、垂直
# 显示图表
plt.show()
```

```python
# 绘制柱状图
import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体
rcParams['font.family'] = 'SimHei'# windows系统
# 创建一个图标，设置大小
plt.figure(figsize=(10, 5))

#添加数据
subjects = ['yuwen','shuxue','waiyu','kexue']
scores = [85,92,78,88]
#绘制柱状图
plt.bar(subjects,scores,
        label= '小明',
        width=0.6)
#添加标题
plt.title('2025学习成绩分布', color='red',fontsize=20)
# 添加坐标轴标签
plt.xlabel('科目',fontsize=40)
plt.ylabel('分数',fontsize=15)
# 添加图例
plt.legend(loc='upper right')
##添加网格线
plt.grid(axis='y',alpha=0.3,color='green',linestyle='--') # 透明度 颜色 虚实线
# plt.grid(axis='y') 只有横格
# plt.grid(axis='x) 只有竖格
# 设置刻度字体大小
plt.xticks(rotation=10,fontsize=20) #旋转角度
plt.yticks(rotation=0,fontsize=15)    #ctrl+/ 一键注释
# 设置y轴的范围
plt.ylim(0,100)
# 数据点显示数值
for x,y in zip(subjects,scores):
    plt.text(x,y,str(y),ha='center',va='bottom',fontsize=13) # 水平、垂直
# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 绘制饼图
import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体
rcParams['font.family'] = 'SimHei'# windows系统
# 创建一个图标，设置大小
plt.figure(figsize=(10, 5))

# #添加数据
things = ['学习', '娱乐','运动','睡觉','其他']
time = [6,3,2,8,5]
pie_colors = ['#66b3ff','#99ff99','#ffcc99','#ff9999','#ff4499'] # 配色
#绘制饼图
plt.pie(time,labels=things,
        autopct='%.2f%%',  # 显示百分比
        startangle=0, # 旋转角度
        colors = pie_colors) 
#添加标题
plt.title('一日时间安排', color='red',fontsize=20)

# 自动优化排版
plt.tight_layout()
# 显示图表
plt.show()
```

```python
# 绘制环形图
import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体
rcParams['font.family'] = 'SimHei'# windows系统
# 创建一个图标，设置大小
plt.figure(figsize=(10, 5))

# #添加数据
things = ['学习', '娱乐','运动','睡觉','其他']
time = [6,3,2,8,5]
pie_colors = ['#66b3ff','#99ff99','#ffcc99','#ff9999','#ff4499'] # 配色
#绘制环形图
plt.pie(time,labels=things,
        autopct='%.2f%%',  # 显示百分比
        startangle=90, # 设置初始旋转角度
        colors = pie_colors, # 设置饼图配色
        wedgeprops = {'width':0.55}, # 设置圆环宽度百分比
        pctdistance=0.70 #设置百分比数字的位置
        )
#添加标题
plt.title('一日时间安排', color='red',fontsize=20)
plt.text(0,0,'总计：\n100%',ha ='center',va ='bottom',fontsize=10)
# 自动优化排版
plt.tight_layout()
# 显示图表
plt.show()
```

```python
# 绘制爆炸式饼图
import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体
rcParams['font.family'] = 'SimHei'# windows系统
# 创建一个图标，设置大小
plt.figure(figsize=(10, 5))

# #添加数据
things = ['学习', '娱乐','运动','睡觉','其他']
time = [6,3,2,8,5]
pie_colors = ['#66b3ff','#99ff99','#ffcc99','#ff9999','#ff4499'] # 配色
explpde =[0.1,0,0,0,0] # 设置突出块的位置
#绘制饼图
plt.pie(time,labels=things,
        autopct='%.2f%%',  # 显示百分比
        startangle=0, # 旋转角度
        colors = pie_colors,
        explode=explpde, # 设置突出块
        shadow=True,  # 阴影
        ) 
#添加标题
plt.title('一日时间安排', color='red',fontsize=20)

# 自动优化排版
plt.tight_layout()
# 显示图表
plt.show()
```

```python
# 绘制散点图
import matplotlib.pyplot as plt
from matplotlib import rcParams # 字体
import random
rcParams['font.family'] = 'SimHei'# windows系统
# 创建一个图标，设置大小
plt.figure(figsize=(10, 8))

#添加数据
x = []
y = []
for i in range(1000):
    tmp = random.uniform(0,10)
    x.append(tmp)
    tmp2 = 2*tmp + random.gauss(0,2)
    y.append(tmp2)
#绘制散点图
plt.scatter(x, y,
            color='blue',
            alpha=0.5,
            s=20,
            label='数据')
#添加标题
plt.title('学习时间与成绩的关系', color='black',fontsize=20)
# 添加坐标轴标签
plt.xlabel('学习时间',fontsize=40)
plt.ylabel('成绩',fontsize=15)
# 添加图例
plt.legend(loc='upper right')
##添加网格线
plt.grid(True,alpha=0.3,color='green',linestyle='--') # 透明度 颜色 虚实线
# plt.grid(axis='y') 只有横格
# plt.grid(axis='x) 只有竖格
# 设置刻度字体大小
plt.xticks(rotation=10,fontsize=20) #旋转角度
plt.yticks(rotation=0,fontsize=15)
# 设置y轴的范围
plt.ylim(0,30)
plt.plot([0,10],[0,20],color='red',linewidth=3,linestyle='-')
# 显示图表
plt.show()
```

```python
#箱线图
import matplotlib.pyplot as plt
#模拟3门课的成绩
data ={
'语文':[82,85, 88, 70, 90, 76, 84, 83, 95],
'数学':[75,80, 79, 93, 88, 82, 87, 89, 92],
'英语':[70, 72, 68, 65,78, 80, 85, 90, 95]
}
plt.figure(figsize=(8, 6))
plt.boxplot(data.values(), tick_labels=data.keys())
plt.title("各科成绩分布(箱线图)")
plt.ylabel("分数")
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# 案例分析：温度分析
# 1.导入库
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

# 2.导入数据
df = pd.read_csv(r'C:\Users\Lenovo\Desktop\课件\代码\data\weather.csv')
display(df.head())
# 2.1 绘制气温的趋势变化图
df['date'] = pd.to_datetime(df['date'])
df = df[df['date'].dt.year==2015]
plt.figure(figsize=(20,10))
plt.plot(df['date'],df['temp_max'], label='最高气温')
plt.plot(df['date'],df['temp_min'], label='最低气温')
plt.title('2015年气温趋势变化图')
plt.xlabel('日期',fontsize=40)
plt.ylabel('温度',fontsize=40)
plt.legend()


```

```python
df['temp_mean'] = (df['temp_max'] + df['temp_min']) /2
df.head()
plt.figure(figsize=(20,10))
plt.plot(df['date'],df['temp_mean'], label='平均气温')
plt.title('2015年平均气温趋势变化图')
plt.xlabel('日期',fontsize=40)
plt.ylabel('温度',fontsize=40)
plt.legend()
```
