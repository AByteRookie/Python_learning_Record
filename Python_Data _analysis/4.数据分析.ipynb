```python
# 分组聚合
# df.groupby('分组的字段')['聚合的字段'].聚合函数
import pandas as pd
df = pd.read_csv(r'C:\Users\Lenovo\Desktop\课件\代码\data\employees.csv')
df = df.dropna(subset=['department_id'])
df['department_id'] = df['department_id'].astype('int64')
# 计算不同部分的平均薪资
df.groupby('department_id').groups # 查看分组情况
df.groupby('department_id').get_group(110) # 查看特定分组数据情况
df2 = df.groupby('department_id')[['salary']].mean().round(2)
df2 = df2.reset_index()
df2.sort_values('salary',ascending=False)
```

```python
# 计算不同部分不同岗位的人的平均薪资
df2 = df.groupby(['department_id','job_id'])[['salary']].mean().round(2)
df2 = df2.reset_index()
df2['salary'] = df2['salary'].round(1)
df2.sort_values('salary',ascending=False)
```

```python
# 企鹅数据分析
# 1.导入必要的库
import pandas as pd
import numpy as np
# 2.导入数据
df = pd.read_csv(r'C:\Users\Lenovo\Desktop\课件\代码\data\penguins.csv')
df.head(5)
df.info()

```

```python
# 3.数据清洗
# 缺失值的检查
print(df.isna().sum())
df.dropna(inplace=True)
```

```python
# 4.数据特征的构造
df['sex'] = df['sex'].astype('category')
df['bill_ratio'] = df['bill_length_mm'] /df['bill_depth_mm']
```

```python
# 5.数据分析
# 数据分箱_body_mass_g
labels = ['low', 'mid', 'high']
df['mass_level'] = pd.cut(df['body_mass_g'],bins =3,labels=labels)
print(df['mass_level'].value_counts())
# 按岛屿、性别分组分析
df.groupby(['sex','island']).agg({
    'body_mass_g':['mean','count'],
})
```

```python
# 1.导入库
import pandas as pd
import numpy as np
# 2.导入数据
df = pd.read_csv(r'C:\Users\Lenovo\Desktop\课件\代码\data\sleep.csv')
df.head()
df.info()
df.describe()
```

```python
# 3.数据清洗
df.isna().sum()
df.drop(columns='sleep_disorder',inplace=True)
```

```python
# 4.数据转换
df['gender'] = df['gender'].astype('category')
df['occupation'] = df['occupation'].astype('category')
df['bmi_category'] = df['bmi_category'].astype('category')
df[['high','low']] = df['blood_pressure'].str.split('/',expand=True)
# 睡眠质量分箱
labels = ['bad','mid','good']
df['quality_level'] = pd.cut(df['sleep_quality'],bins=3,labels=labels)
age_labels = ['青少年', '中年', '老年']
df['age_level'] = pd.cut(df['age'],bins=3,labels=age_labels)
df.head()
```

```python
# 5.数据分析与统计
print(df['bmi_category'].value_counts())
```

```python
# 根据不同的bmi分组，睡眠质量
df.groupby(['age_level','bmi_category']).agg({
    'sleep_duration':'mean',
    'sleep_quality':'mean',
    'stress_level':'mean'
})
```
