```python
# 缺失值的处理
import pandas as pd
import numpy as np
s = pd.Series([12,25,np.nan, None, pd.NA])
df = pd.DataFrame([[12,pd.NA,2],[25,3,5],[None,4,6]],
                  columns=['第一列','第二列','第三列'])
print(s)
# 缺失值的检测
print(s.isna())
print(s.isnull())
print(df.isna())
print(df.isnull())
print(df.isna().sum(axis=1)) #列
print(s.isna().sum())
print('-'*30)
# 剔除缺失值
print(s.dropna())
print(df.dropna())
print(df.dropna(how='all')) 
print(df.dropna(thresh=2))# 至少两个不是缺失值就保留
print(df.dropna(axis=1))# 剔除一整列的记录
print(df.dropna(subset=['第一列']))
```
