# series的创建

```python
import pandas as pd
```

```python
# series的创建
s = pd.Series([10, 2, 3, 4, 5])
# 自定义索引
# s = pd.Series([10, 2, 3, 4, 5], index=['A', 'B', 'C', 'D', 'E'])
# s = pd.Series([10, 2, 3, 4, 5], index=[1, 2, 3, 4, 5])
#定义标签
s = pd.Series([10, 2, 3, 4, 5], index=['A', 'B', 'C', 'D', 'E'], name = '月份')
print(s)
```

```python
# 通过字典的方式创建
s = pd.Series({"a":1,"b":2,"c":3,"d":4,"e":5})
print(s)
s1 = pd.Series(s,index=["a", "c"])
print(s1)
s2 = pd.Series([10, 2, 3, 4, 5], index=['A', 'B', 'C', 'D', 'E'], name = '月份')
s21 = pd.Series(s2,index=["A", "D"])
print(s21)
```

```python
# series的属性
"""index:Series的索引对象
values:Series的值
dtype或dtypes:Series的元素类型
shape:Series的形状
ndim:Series的维度
size:Series的元素个数
name:Series的名称
loc[显式索引，按标签索引或切片
iloc[]隐式索引，按位置索引或切片
at[]使用标签访问单个元素
iat[]使用位置访问单个元素
"""
print(s.index)
print(s.values)
print(s.shape,s.ndim,s.size)
print(s.dtype)
s.name = 'test'
print(s.name)
print(s.loc['a': 'c'])# 显式索引
print(s.iloc[1:4]) # 隐式索引
print(s.at['a']) # 不支持切片
print(s.iat[3])
```

```python
# 访问数据
#print(s['c'])
print(s[s<=3])
s['f'] = 6
print(s.head(2))
print(s.tail(1))
```

```python
#创建一个包含10名学生数学成绩的Serie8，成绩范围在50-100之间。
#计算平均分、最高分、最低分，并找出高于平均分的学生人数。
```

```python
import pandas as pd
import numpy as np
np.random.seed(42)
scores = pd.Series(np.random.randint(50, 101, 10), index=['学生'+str(i) for i in range(1, 11)])
print(scores)
print('平均分：', scores.mean())
print('最高分： ', scores.max())
print('最低分： ', scores.min())
mean = scores.mean()
print('高于平均分的学生人数：', scores[scores>mean].count())
print(len(scores[scores>mean]))
```

```python
#给定某城市一周每天的最高温度Series，完成以下任务:找出温度超过30度的天数
#计算平均温度
#将温度从高到低排序
#找出温度变化最大的两天
```

```python
import pandas as pd
import numpy as np
temperatures = pd.Series([28, 31, 29, 32, 30, 27, 33], # values
               index=['周一','周二','周三','周四','周五','周六','周日'],
               name='温度')
temperature = temperatures>30
print(temperature.to_frame())
print('温度大于30的天数：', len(temperatures[temperatures>30]))
print('平均值：', temperatures.mean())
t2 =  temperatures.sort_values(ascending=False)
print(f"从高到低排序:\n{t2}")
# 温度变化最大的两天
t3 = temperatures.diff().abs() # 计算series数组的变化的值

print('温度变化最大的两天',t3.sort_values(ascending=False).index[:2].to_list())
```

```python
#给定某股票连续10个交易日的收盘价Series:
# 计算每日收益率(当日收盘价/前日收盘价-1)
# 找出收益率最高和最低的日期
# 计算波动率(收益率的标准差)
import pandas as pd
import numpy as np
prices = pd.Series([102.3, 103.5, 105.1, 104.8, 106.2, 107.0, 106.5, 108.1, 109.3, 110.2],
                   index = pd.date_range('2023-01-01', periods=10))
print(prices)
a = prices.pct_change() #percent change （今-前）/前 1
print(a.idxmax())  #日期
print(a.idxmin())  #最低日期
print(a.std())  #波动率
```

```python
#某产品过去12个月的销售量Series:
#计算季度平均销量(每3个月为一个季度)
#找出销量最高的月份
#计算月环比增长率 环比=（今-前）/前  同比=每年同期
# 找出连续增长超过2个月的月份
sales = pd.Series([120, 135, 145, 160, 155, 170, 180, 175, 190, 200, 210,
        220],index=pd.date_range('2022-01-01', periods=12, freq='MS')) #M=月
#季度平均销量
print(sales.resample('QS').mean()) # 重新采样 .sum() .median()
#销量最高的月份
print('销量最高的月份： ', sales.idxmax())
# 月环比增长率
print('月环比增长率')
print(sales.pct_change())
#连续增长超过两个月的月份
a = sales.pct_change()
b =a>0
print(b)
b[b.rolling(3).sum()==3].keys().tolist()
```

```python
# 某商店每小时销售额Series:
# 按天重采样计算每日总销售额
# 计算每天营业时间(8:00-22:00)和非营业时间的销售额比例
# 找出销售额最高的3个小时
import pandas as pd
import numpy as np
np.random.seed(42)
hours_sales=pd.Series(np.random.randint(0, 100, 24),
          index=pd.date_range('2025-01-01', periods=24,freq='h'))
# 按天重采样计算每日总销售额
day_sales = hours_sales.resample('D').sum()
# hours_sales.sum()
# 计算每天营业时间(8:00-22:00)和非营业时间的销售额比例
# hours_sales.between_time('8:00', '22:00') # 筛选一段时间内额series
business_hours_sales = hours_sales[(hours_sales.index.hour>=8) & 
                                   (hours_sales.index.hour<=22)].sum()
print(business_hours_sales / (day_sales-business_hours_sales))
# 找出销售额最高的3个小时
print(hours_sales.nlargest(3))
```
